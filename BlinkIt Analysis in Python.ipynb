### **Data Cleaning**

```python
print(df['Item_Fat_Content'].unique())
```

```python
df['Item_Fat_Content'] = (
    df['Item_Fat_Content']
    .str.strip()
    .replace({
        'LF': 'Low Fat',
        'low fat': 'Low Fat',
        'reg': 'Regular'
    })
)
```

```python
print(df['Item_Fat_Content'].unique())
```

### **Business Requirements**

### **KPI`s Requirements**

```python
#Total Sales
total_sales = df['Item_Outlet_Sales'].sum()

# Average Sales
avg_sales = df['Item_Outlet_Sales'].mean()

# No of Items Sold
num_items = df['Item_Outlet_Sales'].count()


# Number of Unique Items
unique_items = df['Item_Identifier'].nunique()

# Average Item MRP
avg_mrp = df['Item_MRP'].mean()

# Number of Outlets
total_outlets = df['Outlet_Identifier'].nunique()


# Display 
print(f"Total Sales: ${total_sales:,.2f}")
print(f"Average Sales: ${avg_sales:,.2f}")
print(f"Number of Records: {num_items:,}")
print(f"Total Item Records: {num_items:,}")
print(f"Unique Items: {unique_items:,}")
print(f"Average Item MRP: ${avg_mrp:,.2f}")
print(f"Total Outlets: {total_outlets:,}")
```

### **CHART REQUIREMENTS**

### **Total Sales by Fat Content**

```python
sales_by_fat = df.groupby('Item_Fat_Content')['Item_Outlet_Sales'].sum()

plt.figure(figsize=(6,6))

plt.pie(
    sales_by_fat,
    labels=sales_by_fat.index,
    autopct='%1.1f%%',   # Correct format
    startangle=90
)

plt.title('Sales Distribution by Fat Content')
plt.show()
```

```python
import matplotlib.pyplot as plt

sales_by_fat = df.groupby('Item_Fat_Content')['Item_Outlet_Sales'].sum()

sales_by_fat.plot(
    kind='bar',
    color=['skyblue', 'orange']
)

plt.title('Total Sales by Fat Content')
plt.xlabel('Fat Content')
plt.ylabel('Total Sales')
plt.show()
```

### **Total Sales by Item Type**

```python
sales_by_type = (
    df.groupby('Item_Type')['Item_Outlet_Sales']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

plt.bar(sales_by_type.index, sales_by_type.values)

plt.title("Total Sales by Item Type")
plt.xlabel("Item Type")
plt.ylabel("Total Sales")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### **Fat Content Outlet for Total Sales**

```python
grouped = df.groupby(['Outlet_Location_Type','Item_Fat_Content'])['Item_Outlet_Sales'].sum().unstack()
grouped = grouped [['Regular','Low Fat']]

ax = grouped.plot(kind='bar', figsize =(8,5),title ='Outlet Tier by  Item Fat Content')
plt.xlabel('Outlet_Location_Type')
plt.ylabel('Total Sales')
plt.legend(title ='Item Fat Content')
plt.tight_layout()
plt.show()
```

### **Total Sales By Outlet Establishment**

```python
import matplotlib.pyplot as plt

sales_by_year = (
    df.groupby('Outlet_Establishment_Year')['Item_Outlet_Sales']
    .sum()
    .sort_index()
)

plt.figure(figsize=(10, 6))
plt.plot(
    sales_by_year.index,
    sales_by_year.values,
    marker='o',
    linewidth=2
)

plt.title('Total Sales by Outlet Establishment Year')
plt.xlabel('Establishment Year')
plt.ylabel('Total Sales')
plt.grid(True, alpha=0.3)

plt.show()
```

### **Total Sales by OutLet size**

```python
sales_by_size = df.groupby('Outlet_Size')['Item_Outlet_Sales'].sum()

plt.figure(figsize=(4,4))
plt.pie(sales_by_size, labels= sales_by_size.index , autopct='%1f%%' , startangle =90)
plt.title('Outlet Size')
plt.tight_layout()
plt.show()
```

### **Sales by Outlet Location**

```python
sales_by_location = df.groupby('Outlet_Location_Type')['Item_Outlet_Sales'].sum().reset_index()
sales_by_location = sales_by_location.sort_values('Item_Outlet_Sales', ascending=False)

plt.figure(figsize=(8,3)) # Smaller height ,enough width
ax = sns.barplot(x='Item_Outlet_Sales', y= 'Outlet_Location_Type',data= sales_by_location)

plt.title('Total Sales by Outlet Location Type')
plt.xlabel('Total Sales')
plt.ylabel('Outlet Location Type')

plt.tight_layout()
plt.show()
```
